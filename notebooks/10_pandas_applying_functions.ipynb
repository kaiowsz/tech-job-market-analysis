```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_salaries = df[pd.notna(df['salary_year_avg'])].copy()
```

```python
def projected_salary(salary):
    return salary * 1.03

df_salaries['salary_year_inflated'] = df_salaries['salary_year_avg'].apply(projected_salary)

df_salaries[['salary_year_avg', 'salary_year_inflated']]
```

```python
def projected_salary(row):
    if "Senior" in row['job_title_short']:
        return row['salary_year_avg'] * 1.05
    else:
        return row['salary_year_avg'] * 1.03

df_salaries['salary_year_inflated'] = df_salaries.apply(projected_salary, axis=1)

df_salaries[['job_title_short', 'salary_year_avg', 'salary_year_inflated']]
```
